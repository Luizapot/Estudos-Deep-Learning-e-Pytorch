# Operações com tensores

```python
torch3 = torch.arange(10) #LIsta os números de 0 até 10 (inicio, fim, passo)
torch3
```

```python
torch3 = torch3.reshape(2,5) # Muda o formato, aqui no caso para 2 linhas com 5 colunas
torch3
```

```python
torch4 = torch.arange(15)
torch4
```

```python
torch4 = torch4.reshape(5, -1) # O -1 é usando quando não se sabe o tamanho da lista
torch4
```

Modificando o original, também se modifica o alterado


```python
torch5 = torch.arange(10)
torch5
```

```python
torch6 = torch5.reshape(2,5)
torch6
```

```python
torch5[0] = 123 # Primeiro termo
print(torch5)
print(torch6)
```

Slice

```python
torch7 = torch.arange(10)
print(torch7)
```

image.png

```python
torch8 = torch7.reshape(5,2)
print(torch8)

print(torch8[0,0]) # Primeira linha, primeira coluna
print(torch8[0]) #Primeira linha
print(torch8[0,-1]) # Primeira linha, útimo número
print(torch8[:,1]) # Toda linha da segunda coluna toda
print(torch8[:,1:]) # Toda linha da segunda coluna toda, mas no formato
print(torch8[1,2:]) # Erro, não existe terceira coluna
print(torch8[1:3]) # Linhas 1 e 3 (Quando não colocar a virgula, fica no formato original)
```

# Operações matemáticas

```python
tensor_a = torch.tensor([1,2,3,4,5])
tensor_b = torch.tensor([6,7,8,9,10])

tensor_a + tensor_b
```

```python
torch.add(tensor_a, tensor_b)
```

```python
tensor_a - tensor_b
```

```python
torch.sub(tensor_a, tensor_b)
```

```python
tensor_a * tensor_b
```

```python
torch.multiply(tensor_a, tensor_b)
```

```python
tensor_a / tensor_b
```

```python
torch.div(tensor_a, tensor_b)
```

```python
tensor_a % tensor_b
```

```python
torch.remainder(tensor_a, tensor_b)
```

```python
torch.pow(tensor_a, tensor_b)
```

```python
tensor_a.add_(tensor_b) #Adiconar b no a
tensor_a
```
